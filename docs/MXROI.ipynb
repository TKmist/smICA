## Mix ROI
This tool is designed to perform a product, sum, or subtract operations on two ROI files generated with smICA.

### Using the ROI mixer tool
1. From the Tool menu, select **ROI mixer**
```{image} img/mxroi/select_tool.png
    :alt: Corr panel\n",
    :width: 80%\n",
    :align: center"

It will open a new window that will help to navigate through the process,   
```{image} img/mxroi/MX_01.png
    :alt: Corr panel\n",
    :width: 80%\n",
    :align: center"

Using the _ROI1_, _ROI2_ buttons define ROI files source to be used.
```{image} img/mxroi/MX_02.png
    :alt: Corr panel\n",
    :width: 80%\n",
    :align: center"

The _ROI3_ button defines the target ROI folder. The name of the resulting file will be the same as for ROI 1.
```{image} img/mxroi/MX_03.png
    :alt: Corr panel\n",
    :width: 80%\n",
    :align: center"

In the image fields below the buttons the ROI images will appear. First two represents ROI 1 and ROI 2. The third image is a result of operation on the first two ROIS. 
```{image} img/mxroi/MX_04.png
    :alt: Corr panel\n",
    :width: 80%\n",
    :align: center"

In the middle of the window there is a drop down menu with available operations.
```{image} img/mxroi/MX_05.png
    :alt: Corr panel\n",
    :width: 80%\n",
    :align: center"

Press the _Proceed_ button to save ROI 3 into file.
```{image} img/mxroi/MX_06.png
    :alt: Corr panel\n",
    :width: 80%\n",
    :align: center"

The ROI files created with this toola can be directly loaded into the smICA analyis workflow.
