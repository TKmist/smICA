# Supported equipment

This page distinguishes equipment explicitly supported and validated by smICA from data that can be opened through an optional third-party compatibility reader. A device listed by the third-party reader must not be interpreted as having complete, experimentally validated smICA support.


## Explicit smICA support

| Equipment | Acquisition and file requirements | Reader | Supported smICA operations |
|---|---|---|---|
| **PicoHarp 300** | PicoQuant `.ptu` imaging data in **T3** mode with the `rtPicoHarp300T3` record type. The file must describe a FLIM acquisition and contain supported raster geometry together with line-start, line-stop and frame markers. | The smICA reconstruction backend based on `phconvert==0.10.1`. | Header validation, detector routing, raster reconstruction, integration of every complete frame, time gating, TCSPC statistical filtering, weighted lifetime calculation, compact frame-resolved foreground/background storage and subsequent Phot2conc analysis. |

PicoHarp 300 T3 is currently the only hardware and acquisition combination for which smICA makes an explicit support claim. It is covered by the dedicated reader, numerical tests and the end-to-end reference workflow. The claim applies to compatible PTU raster-imaging acquisitions; it does not extend automatically to every scanner, marker configuration, PicoHarp file type or T2 acquisition.


## Third-party PicoQuant compatibility

PTU record types other than `rtPicoHarp300T3` are passed to the optional [`readPTU_FLIM`](https://github.com/TKmist/readPTU_FLIM/tree/NIKON_correction) fallback. This component is maintained outside the smICA repository and may be downloaded during installation. Its documentation declares T2 and T3 record handling for the PicoQuant device families listed below.

| Device family represented by the third-party reader | Declared record modes | Upstream validation statement | smICA status |
|---|---|---|---|
| PicoHarp | T2 and T3 | The upstream project reports FLIM tests with PicoHarp data. | Third-party fallback; configurations other than the dedicated PicoHarp 300 T3 route are not explicitly validated by smICA. |
| HydraHarp, including HydraHarp V2 records | T2 and T3 | The upstream project reports FLIM tests with HydraHarp data. | Third-party fallback; not explicitly validated by smICA. |
| TimeHarp 260 N | T2 and T3 | A decoder is provided, but the upstream README does not include TimeHarp in its list of tested FLIM hardware. | Third-party fallback; not explicitly validated by smICA. |
| TimeHarp 260 P | T2 and T3 | A decoder is provided, but the upstream README does not include TimeHarp in its list of tested FLIM hardware. | Third-party fallback; not explicitly validated by smICA. |
| MultiHarp 150 N | T2 and T3 | The upstream project reports FLIM tests with MultiHarp data. | Third-party fallback; not explicitly validated by smICA. |

The third-party project also declares image reconstruction for laser-beam-scanner (LSM) and piezo-scanner acquisitions. These statements describe the scope of `readPTU_FLIM`, not an independent smICA validation.


## Scope and limitations of the fallback

- A record decoder being present does not guarantee that a particular microscope, scanner, marker layout or detector routing can be reconstructed correctly.
- Although `readPTU_FLIM` declares T2 and T3 record decoding, the smICA Extract and Phot2conc workflow requires reconstructible FLIM/TCSPC image data. T2 decoding alone is therefore not a claim of end-to-end smICA compatibility.
- The fallback supplies the historical integrated image path. It does not expose the dedicated PicoHarp backend's frame-resolved reconstruction used for frame-wise foreground and background exports.
- Historical fallback files retain the established `N-1` effective-frame interpretation caused by the legacy marker logic. Phot2conc applies the matching frame divisor so that the downstream calculation remains internally consistent.
- Third-party compatibility depends on the optional `readPTU_FLIM.py` component being available in the installed environment. Its licence, maintenance and hardware claims belong to the external project.

Before using another device for quantitative work, validate photon totals, image geometry, detector assignment, TCSPC bins, frame markers and concentration results against an independent reference measurement. Representative PTU files and known expected results are required before a new hardware combination can be promoted to explicit smICA support. See [For developers](For_developers.ipynb) for the extension and validation route.
