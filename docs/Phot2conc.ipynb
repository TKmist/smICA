## Phot 2 conc
This mode allows for mapping the concentration of fluorescent molecules using single-molecule FLIM imaging data.
The script requires the _.pkl_ files created in the [EXTRACT from PTU and FILTER](#Extract-from-PTU-and-filter) mode. __Important! This script does not read the _.ptu_ files directly. It is necessary to use the [EXTRACT from PTU and FILTER](#Extract-from-PTU-and-filter) mode first!__

### Open the data directory

To open the folder containing the files for analysis, select **Open PTU directory** from the File menu. The directory may contain PTU files with their extracted PKL products. Historical PKL files remain readable; version-2 PKL files additionally provide integrated background/lifetime maps and sparse frame-resolved foreground/background data.

>  **For version v2.2.0 and higher:**

Phot2conc accepts current PTU/PKL datasets and retains backward compatibility with older PKL data. Frame-resolved export columns are produced only when the loaded PKL contains `frame_data_<channel>` or the supported legacy frame arrays. Missing frame data do not change the integrated calculation.

### The main window 
The GUI is divided into several sections.
1. PTU metadata containing the information about image acquisition.
2. ROI containing the selection panel for choosing the method for the region of interest. It also allows control of the automatic ROI [see below](#Auto-ROI).
3. The list of files for analysis that are located in the PTU folder.
4. Buttons starting the calculations and export options.
5. Analysed image - 5' for channel 1 and 5'' for channel 2.
6. Histograms of the concentration number of molecules or number of photons per pixel - 6' for channel 1 and 6'' for channel 2.
7. The FCS calibration data
8. Mean results.

#### The region of interest
The region of interest, ROI, is a part of the image that will be considered during the image analysis. It is marked in red on the image. The software has two options for ROI definition: [ROI from files](#ROI-from-files) or [Auto ROI](#Auto-ROI).

##### ROI from files
This option requires externally created roi files. Examples of externally generated ROI files (generated with ImageJ) are in the [samples folder on the repository site](https://github.com/TKmist/smICA/tree/main/samples/ROI). Before import, the externally created ROI files should be rewritten to a format readable by the smICA software with the [*REWRITE ROI* tool](#Rewrite-ROI).

**Note that**
 - The ROI file for each channel is created separately.
 - The filename of the ROI file should have the same structure as the PTU file with an additional string at the end of the filename, according to the example given below.    
_file_name.ptu_    
_file_name_roi0_ch_1.dat_ - (ROI file for channel 1)    
_file_name_roi0_ch_2.dat_ - (ROI file for channel 2)

The ROI files will load automatically for the given _.ptu_ file. To import the ROI files, the user must specify the folder containing the files by selecting the path using the **Open ROI directory** option from the **File** menu. It is possible to have many ROI files per _.ptu_ file. The multiple ROI files are identified by numeration in the _roiX_ part of the ROI's file name, where X denotes the number of the ROI file.

##### Auto ROI
Selecting the Auto ROI checkbox displays the ROI control panel for both channels, and the ROI on the image as a red field. 
```{image} img/phot/Phot2Conc_autoroi_control.png
    :alt: Corr panel\n",
    :width: 80%\n",
    :align: center"

Currently, the automatic ROI option offers: 
 - automatic detection of a single cell,
 - automatic detection of more than one cell and selection of the ROI of interest (clicking on the image),
 - copying ROI between channels,
 - finding dark or bright areas,
 - subtracting dark or bright areas, 
 - finding and subtracting many bright spots.

All modes use the OTSU thresholding method. The user can independently adjust the original OTSU threshold by changing the corresponding sliders in each channel.

ROI copying between channels is performed only in the "detect cell" mode. Note that when the checkbox is selected, the thresholding levels can be regulated only in the original channel. For example, selecting "Channel 1->2" will copy the ROI from channel 1 to channel 2. 

The ROI generated in auto-mode will be stored as [ROI files](#ROI-from-files) and can be reused if necessary. For this purpose, within the folder containing the analysed _.ptu_ files, a new **ROI** folder will be created.

>  **For version v2.2.0 and higher:**

All automatic-ROI variants operate on **Foreground intensity**. The channel controls expose the cell threshold, optional multiple-cell handling, ROI mode, object threshold and copy-from-other-channel option where applicable. Selecting another measurement rebuilds the candidates from that measurement's foreground image, so an automatic mask must never be assumed to persist unchanged across files.

#### Calculate the results
To correctly calculate concentration maps, the user must provide or import [calibration data](#Calibration-data) obtained from FCS measurements, including the focal volume width $\omega_0$, the structure parameter $\kappa$, and the mean molecular brightness. 

Pressing the **Calculate single** button calculates and displays results for a single file. Press the **Add to results** button to save the results and retain the Auto ROI settings for the current file. The calculate all button will automatically calculate and store the data for all files, one by one.
```{image} img/phot/Phot2Conc_calculate.png
    :alt: Corr panel\n",
    :width: 80%\n",
    :align: center"

>  **For version v2.2.0 and higher:**

The GUI reports integrated values calculated from all reconstructed frames. Frame-resolved results are export-only and therefore do not add columns or controls to the visible GUI table. For every channel and frame, Phot2conc applies the same active ROI and reports foreground weighted counts, background weighted counts, molecule number, concentration and their corresponding uncertainty/statistical fields.

The mean result will be displayed in tables within the **RESULTS** window on the right-hand side of the screen
```{image} img/phot/Phot2Conc_results_tab.png
    :alt: Corr panel\n",
    :width: 80%\n",
    :align: center"

and presented in the form of per-pixel distributions for concentration, number of molecules - N_p, or number of photons, respectively.
```{image} img/phot/Phot2Conc_results_dist.png
    :alt: Corr panel\n",
    :width: 80%\n",
    :align: center"

The error values displayed in the RESULTS window account for overlap between adjacent confocal observation areas. For a Gaussian lateral profile, pixels separated by distance $d$ are assigned the correlation $\rho(d)=\exp(-d^2/\omega^2)$. The ROI geometry, pixel size and channel-specific $\omega$ are used to calculate the effective sample size $N_{\mathrm{effective}}=N_{\mathrm{pixels}}^2/\sum_{ij}\rho(d_{ij})$. Consequently, the spatial uncertainty is the effective SEM, $SD/\sqrt{N_{\mathrm{effective}}}$, rather than $SD/\sqrt{N_{\mathrm{pixels}}}$.

The checkbox _Errors as SEM_ switches between two reported uncertainties:
 - With the checkbox unmarked (default), `N_p_err` and `C_err` combine the effective SEM with the shared calibration uncertainty in quadrature. Molecular-number calibration uncertainty originates from molecular brightness; concentration calibration uncertainty includes molecular brightness and effective focal volume.
 - With the checkbox marked, `N_p_err` and `C_err` contain only the effective SEM. Calibration uncertainty is not included.

The exported `N_effective_pixels` column records the effective sample size used for the integrated result. It approaches the physical ROI pixel count only when overlap correlations are negligible; equality of pixel size and $\omega$ is not sufficient because nearest-neighbour correlation is then $e^{-1}$.

#### Export results
During calculation, the data will be automatically exported according to the rules marked in the export panel.
```{image} img/phot/Phot2Conc_results_export.png
    :alt: Corr panel\n",
    :width: 80%\n",
    :align: center"

>  **For version v2.2.0 and higher:**

Frame-resolved fields use one-based names of the form `Frame_<N>_<metric>`. Available metrics are `<Counts>`, `Counts_std`, `<N_p>`, `N_p_err`, `<C>`, `C_err`, `C_median`, `C_median_abs_err`, `<Background counts>`, `Background_counts_std` and `N_effective_pixels`. Counts are TCSPC-weighted counts per pixel in the active ROI, `N_p` is molecules per pixel and `C` is concentration in nM. Frame 1 is the first reconstructed frame.

These columns are appended only to exported CSV, DAT or pickle tables. Fractional foreground/background counts are expected after statistical filtering because they are sums of event weights rather than raw photon counts. Older PKL files without frame products export only integrated fields.

The **to array** means that the script will create a _.csv_ file containing the array (size of the image) containing a number of photons, molecules, or concentration in a given pixel.    
"to heatmap" means that the heatmap will be exported as the _.png_ file.   

To export the averaged data as a table, press the "Export all data" button.
