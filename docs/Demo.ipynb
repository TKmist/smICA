## Bundled end-to-end demo

smICA includes a self-contained demonstration that runs the complete workflow from PTU extraction through calibration loading and concentration calculation. The same Python automation script is used on Linux and Windows; only the launcher command differs.

```bash
# Linux
./run_smICA --demo
```

```bat
:: Windows
smICA.bat --demo
```

The launcher starts a dedicated local automation server, waits until the GUI is ready, sends every analysis command in a documented sequence, exports the results, and closes that dedicated smICA process. Keep the GUI untouched until the terminal reports that the demo has completed.

### Included measurements

The installed `smICA/calc_tests/demo_data` directory contains exactly four PTU files and the required calibration JSON files:

- `in_vitro/dextran_60nm`: one moderate-concentration 60 nM dextran measurement;
- `in_vitro/dextran_80nm`: one moderate-concentration 80 nM dextran measurement;
- `in_cellulo/cells_without_mrna/imaging_new_ptu`: one dual-channel cell measurement without mRNA and its session calibration;
- `in_cellulo/cells_without_mrna/imaging2_17_07_23`: a second dual-channel cell measurement without mRNA and its separate session calibration.

The dextran measurements are extracted without statistical filtering and calculated without an ROI. The cell measurements come from the `pie_dual_channel` reference datasets. Both cellular channels are extracted with TCSPC statistical background filtering and then use their established dual-channel automatic ROI configurations. For each cellular channel, the automation enables statistical filtering, configures the automatic background range, calculates and accepts the filters, and selects the current-decay foreground filter. The selected `imaging_new_ptu` measurement is additionally re-extracted with the standard PIE time gates and statistical filtering disabled. The two variants are calculated with the same calibration and ROI settings and exported separately as `results_tcspc.csv` and `results_gating.csv`; the other datasets export `results.csv`. Generated `.pkl` intermediates remain beside the PTU files so that the extraction result can be inspected or reused.

The much larger `calc_tests/REFERENCE_DATA` collection is intentionally excluded from release archives. It is used only for repository-level reference validation and is not required by `--demo`.

### Automation sequence

The readable source of the workflow is `calc_tests/demo_data/run_demo.py`. Dataset operations are written as explicit `send_command(...)` calls rather than hidden in dataset loops. This makes the TCSPC-filter and time-gating configurations, ROI thresholds, calibration paths, calculations, and exports directly auditable. If any server command fails, the demo stops, reports that command and returns a non-zero exit status.
