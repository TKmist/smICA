## Rewrite ROI
This tool is designed to rewrite ROI files from the format generated by ImageJ into a format readable by the scripts in the Phot2Conc folder.

### Prepare of the ROI files
1. Start ImageJ or Fiji
2. Open one of the  _.png_ file obtained within the [EXTRACT from PTU and FILTER](#Extract-from-PTU-and-filter) mode, either garyscale or false-colours; see the [OUTPUT section](PTUExtract.ipynb#Output) from the [EXTRACT from PTU and FILTER](#Extract-from-PTU-and-filter).
3. Make a selection around the region of interest,
4. In the Edit menu, select _Clear Outside_,
5. In the Edit menu, select _Fill_,
6. Save the image as _Text Image_.
