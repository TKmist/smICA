# Analytical workflow for sample file analysis

> Each terminal data directory listed below contains one calibration JSON file and only the PTU files that use that calibration. Process every directory separately.

## Extract from PTU

### Without filtering

#### Standard-mode data

1. Run smICA in the **Extract from PTU** mode.
2. Open the first directory from the list below.
3. Disable **Remove additional lines** correction.
4. Press **Apply to all** to extract and read all files.
5. Repeat steps 2–4 for every remaining directory:
   - `calc_tests/REFERENCE_DATA/TypeOfFiltering/raw/standard_2022_10_frames/5uW`
   - `calc_tests/REFERENCE_DATA/TypeOfFiltering/raw/standard_2022_10_frames/10uW`
   - `calc_tests/REFERENCE_DATA/TypeOfFiltering/raw/standard_2022_10_frames/20uW`
   - `calc_tests/REFERENCE_DATA/TypeOfFiltering/raw/standard_2022_10_frames/40uW`
   - `calc_tests/REFERENCE_DATA/TypeOfFiltering/raw/standard_2025_frame_series/19_09_60`
   - `calc_tests/REFERENCE_DATA/TypeOfFiltering/raw/standard_2025_frame_series/19_09_80`
   - `calc_tests/REFERENCE_DATA/TypeOfFiltering/raw/standard_2025_frame_series/19_09_100`
   - `calc_tests/REFERENCE_DATA/TypeOfFiltering/raw/standard_2025_frame_series/22_09_60`
   - `calc_tests/REFERENCE_DATA/TypeOfFiltering/raw/standard_2025_frame_series/22_09_100`

#### PIE dual-channel data

1. Open the first directory from the list below.
2. Enable **Remove additional lines** correction.
3. For Channel 1, set **Time>** to 0.
4. For Channel 2, set **Time<** to 50.
5. Press **Apply to all**.
6. Repeat steps 1–5 for the second directory:
   - `calc_tests/REFERENCE_DATA/TypeOfFiltering/raw/pie_dual_channel_128x128_100_frames/imaging_new_ptu`
   - `calc_tests/REFERENCE_DATA/TypeOfFiltering/raw/pie_dual_channel_128x128_100_frames/imaging2_17_07_23`

#### PIE GFP data

1. Open `calc_tests/REFERENCE_DATA/TypeOfFiltering/raw/pie_gfp_256x256_200_frames/shared_pie_calibration`.
2. Enable **Remove additional lines** correction.
3. For Channel 2, set **Time<** to 50.
4. Press **Apply to all**.

#### PIE mRNA data

1. Open `calc_tests/REFERENCE_DATA/TypeOfFiltering/raw/pie_mrna_256x256_200_frames/shared_pie_calibration`.
2. Enable **Remove additional lines** correction.
3. For Channel 1, set **Time>** to 0.
4. Press **Apply to all**.

### TCSPC filtering

#### Standard-mode data

1. Open the first directory from the list below.
2. Disable **Remove additional lines** correction.
3. Select **Use statistical filters**.
4. Press **Calculate filters**.
5. In the Lifetime background filtering window, select **Set data range for background**.
6. Press **Calculate filters**, and then press **Accept**.
7. Press **Apply to all**.
8. Repeat steps 1–7 for every remaining directory:
   - `calc_tests/REFERENCE_DATA/TypeOfFiltering/tcscp/standard_2022_10_frames/5uW`
   - `calc_tests/REFERENCE_DATA/TypeOfFiltering/tcscp/standard_2022_10_frames/10uW`
   - `calc_tests/REFERENCE_DATA/TypeOfFiltering/tcscp/standard_2022_10_frames/20uW`
   - `calc_tests/REFERENCE_DATA/TypeOfFiltering/tcscp/standard_2022_10_frames/40uW`
   - `calc_tests/REFERENCE_DATA/TypeOfFiltering/tcscp/standard_2025_frame_series/19_09_60`
   - `calc_tests/REFERENCE_DATA/TypeOfFiltering/tcscp/standard_2025_frame_series/19_09_80`
   - `calc_tests/REFERENCE_DATA/TypeOfFiltering/tcscp/standard_2025_frame_series/19_09_100`
   - `calc_tests/REFERENCE_DATA/TypeOfFiltering/tcscp/standard_2025_frame_series/22_09_60`
   - `calc_tests/REFERENCE_DATA/TypeOfFiltering/tcscp/standard_2025_frame_series/22_09_100`

#### PIE mRNA and GFP data

1. Open the first directory from the list below.
2. Enable **Remove additional lines** correction.
3. Select **Use statistical filters**.
4. Calculate and accept the lifetime background filter.
5. Press **Apply to all**.
6. Repeat steps 1–5 for the second directory:
   - `calc_tests/REFERENCE_DATA/TypeOfFiltering/tcscp/pie_mrna_256x256_200_frames/shared_pie_calibration`
   - `calc_tests/REFERENCE_DATA/TypeOfFiltering/tcscp/pie_gfp_256x256_200_frames/shared_pie_calibration`

#### PIE dual-channel data

1. Open the first directory from the list below.
2. Enable **Remove additional lines** correction.
3. Select **Use statistical filters** for Channel 1.
4. Calculate and accept the lifetime background filter.
5. Repeat steps 3–4 for Channel 2.
6. Press **Apply to all**.
7. Repeat steps 1–6 for the second directory:
   - `calc_tests/REFERENCE_DATA/TypeOfFiltering/tcscp/pie_dual_channel_128x128_100_frames/imaging_new_ptu`
   - `calc_tests/REFERENCE_DATA/TypeOfFiltering/tcscp/pie_dual_channel_128x128_100_frames/imaging2_17_07_23`
